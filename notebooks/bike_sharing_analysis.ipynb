```python
# Correlation + VIF

import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor
continuous = ['temp', 'atemp', 'hum', 'windspeed', 'cnt']
predictors = ['temp', 'atemp', 'hum', 'windspeed']
corr_pearson = hour[continuous].corr('pearson')
corr_spearman = hour[continuous].corr('spearman')
X_vif = sm.add_constant(hour[predictors])
vif = pd.DataFrame({
    'Variable': predictors,
    'VIF': [variance_inflation_factor(X_vif.values, i+1) for i in range(len(predictors))]
}).round(2)
fig, axes = plt.subplots(1, 3, figsize=(13, 4))
sns.heatmap(corr_pearson, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, center=0, square=True, annot_kws={"size": 9}, cbar=False, ax=axes[0])
axes[0].set_title("Pearson", fontsize=12)

sns.heatmap(corr_spearman, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, center=0, square=True, annot_kws={"size": 9}, cbar=False, ax=axes[1])
axes[1].set_title("Spearman", fontsize=12)
# VIF table
axes[2].axis('off')
table = axes[2].table(cellText=vif.values,colLabels=vif.columns,cellLoc='center',loc='center',colWidths=[0.65, 0.35]
)
table.auto_set_font_size(False)
table.set_fontsize(11)
table.scale(1.15, 2.0)
axes[2].set_title("Multicollinearity Check", fontsize=12, pad=15)
plt.tight_layout()
plt.show()
```

## Non-Linearity Check 

```python
# NON-LINEAR STRUCTURE ANALYSIS
from statsmodels.formula.api import ols
# date is datetime
hour['dteday'] = pd.to_datetime(hour['dteday'])
continuous_features = ['temp', 'hum', 'windspeed']
results = []
# Create plots 
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
for i, feature in enumerate(continuous_features):
    # Linear model
    linear_model = ols(
        f'cnt ~ {feature}',
        data=hour
    ).fit()
    # Quadratic model
    quadratic_model = ols(
        f'cnt ~ {feature} + I({feature}**2)',
        data=hour
    ).fit()
    # Store all statistical information
    results.append({
        'Variable': feature,'Linear R²': linear_model.rsquared, 'Quadratic R²': quadratic_model.rsquared, 'Linear AIC': linear_model.aic,'Quadratic AIC': quadratic_model.aic
    })
    # Create quantile bins
    temp_df = hour[[feature, 'cnt']].copy()

    temp_df['bin'] = pd.qcut(
        temp_df[feature],q=10, duplicates='drop'
    )

    binned = temp_df.groupby(
        'bin',
        observed=True
    )['cnt'].mean()
    # Plot
    axes[i].plot(
        range(len(binned)),
        binned.values,
        marker='o'
    )
    axes[i].set_xlabel(f'{feature} (10 quantile bins)')
    axes[i].set_ylabel('Mean rental count')
    axes[i].set_title(f'{feature} vs Rental Demand')
    axes[i].set_xticks(range(len(binned)))
    axes[i].tick_params(axis='x', rotation=45)
    axes[i].grid(alpha=0.3)

plt.tight_layout()
plt.show()

# TABLE WITH ALL MODEL INFORMATION

results_df = pd.DataFrame(results)

display(
    results_df.style.format({
        'Linear R²': '{:.4f}', 'Quadratic R²': '{:.4f}', 'Linear AIC': '{:.2f}', 'Quadratic AIC': '{:.2f}'
    })
)
```

## Statistical Analysis of Binary variable (workingday,holiday and year)


```python
# WELCH'S T-TEST FOR BINARY VARIABLES
from scipy.stats import ttest_ind
import pandas as pd
results = []
# 1. Workingday vs cnt
working = hour.loc[hour['workingday'] == 1, 'cnt']
non_working = hour.loc[hour['workingday'] == 0, 'cnt']
w = ttest_ind(
    working, non_working, equal_var=False)
results.append({
    'Variable': 'Workingday','Group 1': 'Working day','Group 2': 'Non-working day','Welch t': w.statistic,'p-value': w.pvalue
})

# 2. Holiday vs cnt
holiday_0 = hour.loc[hour['holiday'] == 0, 'cnt']
holiday_1 = hour.loc[hour['holiday'] == 1, 'cnt']

welch_holiday = ttest_ind(
    holiday_0,holiday_1,equal_var=False)

results.append({
    'Variable': 'Holiday','Group 1': 'Non-holiday','Group 2': 'Holiday','Welch t': welch_holiday.statistic,'p-value': welch_holiday.pvalue})

# 3. Year vs cnt
year_2011 = hour.loc[hour['yr'] == 0, 'cnt']
year_2012 = hour.loc[hour['yr'] == 1, 'cnt']
welch_year = ttest_ind( year_2011,year_2012, equal_var=False)

results.append({
    'Variable': 'Year', 'Group 1': '2011', 'Group 2': '2012', 'Welch t': welch_year.statistic, 'p-value': welch_year.pvalue})

# FINAL TABLE

welch_results = pd.DataFrame(results)
display(
    welch_results.style.format({
        'Welch t': '{:.3f}',
        'p-value': '{:.3g}'
    })
)
```

# Task 3 :Demand Forecasting & Predictive Modeling

##  FEATURE PREPARATION

```python
target = 'cnt'
base_features = [
    'hr', 'workingday', 'weekday', 'mnth', 'season', 'weathersit',
    'temp', 'hum', 'windspeed', 'holiday', 'yr'
]

categorical_features = ['hr', 'weekday', 'mnth', 'season', 'weathersit']
binary_features = ['workingday', 'holiday', 'yr']
continuous_features = ['temp', 'hum', 'windspeed']

excluded_features = ['casual', 'registered', 'atemp'] 

print("Target:", target)
print("Base Features:", base_features)
print("Excluded Features:", excluded_features)
```

## TEMPORAL TRAIN / TEST SPLIT

```python
# Days 1-20 of each month -> train, Days 21-end -> test

df = hour.copy()
df['dteday'] = pd.to_datetime(df['dteday'])
df['datetime'] = df['dteday'] + pd.to_timedelta(df['hr'], unit='h')
df = df.sort_values('datetime').reset_index(drop=True)
df['day_of_month'] = df['dteday'].dt.day

def temporal_split(data):
    train = data.loc[data['day_of_month'] <= 20].copy()
    test = data.loc[data['day_of_month'] >= 21].copy()
    return train, test

train_df, test_df = temporal_split(df)

print("Training observations:", len(train_df))
print("Testing observations :", len(test_df))
print("\nTraining period:", train_df['datetime'].min(), "to", train_df['datetime'].max())
print("Testing period:", test_df['datetime'].min(), "to", test_df['datetime'].max())
```

## Statistical Count-Regression Baseline

```python
# ============================================================
# NEGATIVE BINOMIAL GLM BASELINE
# ============================================================

import statsmodels.formula.api as smf
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Statistical baseline uses only the original/base features
nb_formula = (
    'cnt ~ temp + hum + windspeed '
    '+ C(hr) + C(weekday) + C(mnth) + C(season) + C(weathersit) '
    '+ workingday + holiday + yr'
)

# Fit NB-GLM only on the training data
nb_model = smf.glm(
    formula=nb_formula,
    data=train_df,
    family=sm.families.NegativeBinomial()
).fit()

# Predict on the held-out test period
nb_prediction = np.clip(
    nb_model.predict(test_df),
    0,
    None
)

# Evaluate
nb_mae = mean_absolute_error(test_df[target], nb_prediction)
nb_rmse = np.sqrt(mean_squared_error(test_df[target], nb_prediction))
nb_r2 = r2_score(test_df[target], nb_prediction)

nb_results = pd.DataFrame([{
    'Model': 'Negative Binomial GLM',
    'MAE': nb_mae,
    'RMSE': nb_rmse,
    'R²': nb_r2
}])

print("Negative Binomial GLM Baseline:")
display(nb_results.round(4))
```

## TRAINING-SIDE FEATURE ENGINEERING (leak-safe)

```python
is_train = df['day_of_month'] <= 20
cnt_train_safe = df['cnt'].where(is_train)

lag_specs_hours = {'lag_24h': 24, 'lag_168h': 168}
roll_specs_hours = {'roll_mean_24h': 24, 'roll_mean_168h': 168}

for name, h in lag_specs_hours.items():
    df[name] = cnt_train_safe.shift(h)
for name, w in roll_specs_hours.items():
    df[name] = cnt_train_safe.shift(1).rolling(w).mean()

lag_cols = list(lag_specs_hours.keys()) + list(roll_specs_hours.keys())
print("Training-side temporal features created:", lag_cols)

model_train = df.loc[is_train].dropna(subset=lag_cols).copy()
```

## XGBoost Model Development

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

def create_xgb():
    return XGBRegressor(
        n_estimators=500, max_depth=6, learning_rate=0.05,
        subsample=0.8, colsample_bytree=0.8,
        objective='reg:squarederror', random_state=42, n_jobs=-1
    )

def recursive_forecast(model, feature_cols, df_all):
    df_sorted = df_all.sort_values('datetime').reset_index(drop=True)
    is_train_row = df_sorted['day_of_month'] <= 20

    known = pd.Series(
        np.where(is_train_row, df_sorted['cnt'].values, np.nan),
        index=df_sorted['datetime'].values
    )

    test_rows = df_sorted.loc[~is_train_row].sort_values('datetime')
    preds = pd.Series(index=test_rows.index, dtype=float)

    for idx, row in test_rows.iterrows():
        dt = row['datetime']
        feat = {c: row[c] for c in base_features}

        if 'lag_24h' in feature_cols:
            feat['lag_24h'] = known.get(dt - pd.Timedelta(hours=24), np.nan)
        if 'lag_168h' in feature_cols:
            feat['lag_168h'] = known.get(dt - pd.Timedelta(hours=168), np.nan)
        if 'roll_mean_24h' in feature_cols:
            window = known.loc[dt - pd.Timedelta(hours=24): dt - pd.Timedelta(hours=1)]
            feat['roll_mean_24h'] = window.mean()
        if 'roll_mean_168h' in feature_cols:
            window = known.loc[dt - pd.Timedelta(hours=168): dt - pd.Timedelta(hours=1)]
            feat['roll_mean_168h'] = window.mean()

        X_row = pd.DataFrame([feat])[feature_cols]
        pred = float(np.clip(model.predict(X_row)[0], 0, None))

        known[dt] = pred
        preds.loc[idx] = pred

    return preds.reindex(test_rows.index)
```

## EXPERIMENTS

```python
experiment_features = {
    'Xgboost_Base':                base_features,
    'Lag-24 only':          base_features + ['lag_24h'],
    'Lag-168 only':         base_features + ['lag_168h'],
    'Rolling-24 only':      base_features + ['roll_mean_24h'],
    'Rolling-168 only':     base_features + ['roll_mean_168h'],
    'All Lags (24+168)':    base_features + ['lag_24h', 'lag_168h'],
    'All Rolling (24+168)': base_features + ['roll_mean_24h', 'roll_mean_168h'],
}

test_df_sorted = df.loc[df['day_of_month'] >= 21].sort_values('datetime')
y_test = test_df_sorted[target]

xgb_results = []
xgb_predictions = {}

for experiment_name, features in experiment_features.items():
    y_train = model_train[target]
    model = create_xgb()
    model.fit(model_train[features], y_train)

    if experiment_name == 'Xgboost_Base':
        X_test = test_df_sorted[features]
        prediction = np.clip(model.predict(X_test), 0, None)
        prediction = pd.Series(prediction, index=test_df_sorted.index)
    else:
        prediction = recursive_forecast(model, features, df)

    prediction = prediction.reindex(test_df_sorted.index)
    xgb_predictions[experiment_name] = prediction

    xgb_results.append({
        'Experiment': experiment_name,
        'MAE': mean_absolute_error(y_test, prediction),
        'RMSE': np.sqrt(mean_squared_error(y_test, prediction)),
        'R²': r2_score(y_test, prediction)
    })

xgb_results_df = pd.DataFrame(xgb_results).sort_values('RMSE').reset_index(drop=True)
print("XGBoost temporal feature experiments (honest multi-step forecast):")
display(xgb_results_df.round(4))

#  NEGATIVE BINOMIAL GLM TO FINAL MODEL COMPARISON

nb_result = pd.DataFrame([{
    'Experiment': 'Negative Binomial GLM','MAE': nb_mae,'RMSE': nb_rmse, 'R²': nb_r2
}])

final_results_df = pd.concat(
    [xgb_results_df, nb_result],
    ignore_index=True
)

final_results_df = (
    final_results_df
    .sort_values('RMSE')
    .reset_index(drop=True)
)
print("Final Model Comparison:")
display(final_results_df.round(4))
```
